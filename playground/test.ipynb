# The way code gets run

1. Jupyter reads the cell and confirm that its a "code" cell and notices when you press run
2. It then goes to jupyter server which opens python kernel
     (Kernal is basically a python process that is alive in the background)
3. Kernel runs the python code and sends it back here
4. Then the code output gets displayed below the code

_"run", loads the python processor, loads installed packages in the background, creates a communication channel, connects the notebook kernel, this environment setup is slow, runs the code and sends it back here and then its displayed_
